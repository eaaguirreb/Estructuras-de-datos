# Teoría de Pilas y Colas - Estructura de Datos

Este cuaderno contiene la explicación teórica, operaciones principales, comparativa e implementación en Python de las estructuras de datos **Pilas** (*Stacks*) y **Colas** (*Queues*).

## 1. Pilas (*Stacks*)

Una **pila** sigue el principio **LIFO** (*Last In, First Out*): **el último elemento en entrar es el primero en salir**.

### Operaciones Principales:
- `push(elem)`: Agrega un elemento a la cima (*top*).
- `pop()`: Elimina y retorna el elemento de la cima.
- `peek()` / `top()`: Consulta el elemento superior sin eliminarlo.
- `isEmpty()`: Verifica si la pila no tiene elementos.

### Casos de Uso:
- Historial del navegador (botón "Atrás").
- Función "Deshacer" (*Undo* / `Ctrl+Z`) en editores de texto.
- Gestión de llamadas a funciones (*Call Stack* y recursión).
- Evaluación y conversión de expresiones matemáticas (Infix a Postfix).

```python
# Implementación de Pila en Python usando list
pila = []

# Push (Insertar elementos)
pila.append(10)
pila.append(20)
pila.append(30)

print(f"Estado inicial de la pila: {pila}")

# Pop (Extraer elemento de la cima)
elemento = pila.pop()  # Retorna 30
print(f"Elemento retirado (cima): {elemento}")
print(f"Estado de la pila tras pop: {pila}")
```

## 2. Colas (*Queues*)

Una **cola** sigue el principio **FIFO** (*First In, First Out*): **el primer elemento en entrar es el primero en salir**.

### Operaciones Principales:
- `enqueue(elem)`: Agrega un elemento al final (*rear* / *back*).
- `dequeue()`: Elimina y retorna el primer elemento (*front*).
- `front()` / `peek()`: Consulta el primer elemento sin eliminarlo.
- `isEmpty()`: Verifica si la cola está vacía.

### Casos de Uso:
- Colas de impresión.
- Planificación de procesos en sistemas operativos (Scheduler CPU).
- Algoritmo de búsqueda en anchura (*Breadth-First Search* - BFS) en grafos.
- Buffers de transmisión de datos en redes o streaming de video.

```python
# Implementación de Cola en Python usando collections.deque
from collections import deque

cola = deque()

# Enqueue (Insertar elementos al final)
cola.append("Tarea A")
cola.append("Tarea B")
cola.append("Tarea C")

print(f"Estado inicial de la cola: {list(cola)}")

# Dequeue (Extraer el primer elemento)
tarea = cola.popleft()  # Retorna "Tarea A"
print(f"Procesando primer elemento: {tarea}")
print(f"Estado de la cola tras dequeue: {list(cola)}")
```

## Comparativa Directa

| Característica | Pila (*Stack*) | Cola (*Queue*) |
| :--- | :--- | :--- |
| **Principio** | **LIFO** (*Last In, First Out*) | **FIFO** (*First In, First Out*) |
| **Punto de inserción** | Cima (*Top*) | Final (*Rear* / *Tail*) |
| **Punto de extracción** | Cima (*Top*) | Frente (*Front* / *Head*) |
| **Complejidad $O(1)$** | Inserción y eliminación | Inserción y eliminación |

# Pilas y Colas en Ciencia de Datos e Inteligencia Artificial

Este cuaderno explora el papel clave de las estructuras de datos **Pilas** (*Stacks*) y **Colas** (*Queues*) en flujos de trabajo de Ciencia de Datos (Data Science) e Inteligencia Artificial (IA), incluyendo ejemplos prácticos en Python listos para ejecutar en Google Colab.

## 1. Aplicaciones Teóricas en Ciencia de Datos e IA

### A. Usos de las Pilas (*Stacks* - LIFO)
1. **Backtracking en Búsqueda y Optimización (IA):** Algoritmos como *Depth-First Search* (DFS) o resolución de laberintos/puzles usan pilas para retroceder cuando una rama del espacio de estados no lleva a la solución.
2. **Evaluación de Grafos de Cómputo (Diferenciación Automática):** Marcos como PyTorch o TensorFlow administran el pase hacia adelante (*forward pass*) y hacia atrás (*backpropagation*) mediante pilas de operaciones para calcular gradientes.
3. **Análisis Sintáctico y Tokenización (NLP):** Verificación de estructuras anidadas (llaves, paréntesis, etiquetas HTML/XML/JSON) en preparación de datos no estructurados.
4. **Deshacer/Rehacer en Herramientas de Anotación:** Gestión del historial de acciones en etiquetado de imágenes o texto.

---

### B. Usos de las Colas (*Queues* - FIFO)
1. **Búsqueda en Amplitud / Anchura (BFS en IA):** Exploración por niveles en grafos de decisión, árboles de juegos y problemas de rutas mínimas.
2. **Web Crawlers y Scraping de Datos:** En ingeniería de datos, los rastreadores web almacenan las URLs descubiertas en colas FIFO para procesarlas ordenadamente.
3. **Procesamiento de Flujos de Datos (Streaming / Ingestion):** Arquitecturas tipo Producer-Consumer donde se procesan eventos en el orden exacto de llegada (mensajería con RabbitMQ, Kafka o `queue.Queue`).
4. **Replay Buffers en Aprendizaje por Refuerzo (Reinforcement Learning):** Colas circulares o de tamaño fijo (`deque(maxlen=N)`) para almacenar transiciones `(estado, acción, recompensa, siguiente_estado)` y tomar muestras para el entrenamiento.

## 2. Ejemplos Prácticos de Código (Colab)

### Ejemplo 1: Pila para Backtracking en IA (Depth-First Search - DFS)
Uso de una pila para explorar un grafo de decisiones o un árbol de estados sin usar recursión explícita.

```python
def dfs_pila(grafo, inicio):
    pila = [inicio]
    visitados = set()
    recorrido = []

    while pila:
        # Extraer el último elemento ingresado (LIFO)
        nodo = pila.pop()

        if nodo not in visitados:
            visitados.add(nodo)
            recorrido.append(nodo)

            # Agregar vecinos a la pila
            for vecino in reversed(grafo.get(nodo, [])):
                if vecino not in visitados:
                    pila.append(vecino)

    return recorrido

# Grafo de ejemplo (Red de decisiones o estados)
grafo_ia = {
    'Estado_Inicial': ['Opcion_A', 'Opcion_B'],
    'Opcion_A': ['Sub_A1', 'Sub_A2'],
    'Opcion_B': ['Sub_B1'],
    'Sub_A1': [],
    'Sub_A2': [],
    'Sub_B1': []
}

resultado_dfs = dfs_pila(grafo_ia, 'Estado_Inicial')
print("Exploración DFS (Pila - LIFO):", resultado_dfs)
```

### Ejemplo 2: Cola de Prioridad / Cola BFS para Web Scraping / Crawling
Exploración por niveles (BFS - FIFO) para scraping y extracción masiva de datos web.

```python
from collections import deque

def web_crawler_bfs(url_semilla, paginas_web, max_paginas=5):
    cola_urls = deque([url_semilla])
    visitadas = set()
    paginas_procesadas = []

    while cola_urls and len(visitadas) < max_paginas:
        # Extraer la primera URL en cola (FIFO)
        url_actual = cola_urls.popleft()

        if url_actual not in visitadas:
            visitadas.add(url_actual)
            paginas_procesadas.append(url_actual)

            # Obtener enlaces salientes e ingresarlos a la cola
            enlaces = paginas_web.get(url_actual, [])
            for enlace in enlaces:
                if enlace not in visitadas:
                    cola_urls.append(enlace)

    return paginas_procesadas

# Simulación de una estructura web para ingesta de datos
mapa_web = {
    'https://site.com': ['https://site.com/data1', 'https://site.com/data2'],
    'https://site.com/data1': ['https://site.com/data3'],
    'https://site.com/data2': ['https://site.com/data4'],
    'https://site.com/data3': [],
    'https://site.com/data4': []
}

urls_ingestadas = web_crawler_bfs('https://site.com', mapa_web)
print("URLs procesadas en orden FIFO (BFS):", urls_ingestadas)
```

### Ejemplo 3: Experience Replay Buffer (Cola de Capacidad Fija) para Reinforcement Learning
En algoritmos de Aprendizaje por Refuerzo (como Deep Q-Networks - DQN), se utiliza una cola con límite de capacidad para guardar las experiencias recientes del agente.

```python
import random
from collections import deque

class ReplayBuffer:
    def __init__(self, capacidad):
        # deque con maxlen actúa como una cola circular de tamaño fijo
        self.buffer = deque(maxlen=capacidad)

    def agregar_experiencia(self, estado, accion, recompensa, siguiente_estado):
        self.buffer.append((estado, accion, recompensa, siguiente_estado))

    def obtener_muestra(self, tamano_lote):
        return random.sample(self.buffer, tamano_lote)

    def __len__(self):
        return len(self.buffer)

# Inicializar buffer con capacidad máxima de 5 elementos
memory = ReplayBuffer(capacidad=5)

# Simular interacción del agente con el entorno
for t in range(1, 8):
    memory.agregar_experiencia(
        estado=f"E{t}",
        accion=f"A{t}",
        recompensa=t * 10,
        siguiente_estado=f"E{t+1}"
    )
    print(f"Paso {t} | Tamaño del buffer: {len(memory)} | Elementos: {[e[0] for e in memory.buffer]}")

# Extraer un lote aleatorio para el entrenamiento del modelo
lote_entrenamiento = memory.obtener_muestra(tamano_lote=3)
print("\nLote de mini-batch muestreado para DQN:", lote_entrenamiento)
```

### Parte 3: Reto de Aplicación (Verificador de Paréntesis)

```python
def verificar_parentesis(cadena):
    pila = []
    for caracter in cadena:
        if caracter == '(':
            pila.append(caracter)
        elif caracter == ')':
            if len(pila) == 0:
                return False
            pila.pop()

    return len(pila) == 0

# Pruebas
print("Prueba '((()))':", verificar_parentesis("((()))"))  # Debería ser True
print("Prueba '(()':", verificar_parentesis("(()"))        # Debería ser False
print("Prueba ')(':", verificar_parentesis(")("))         # Debería ser False
```
